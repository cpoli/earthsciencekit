```python
import numpy as np

import earthsciencekit
from earthsciencekit.units import convert

print(earthsciencekit.__version__)
```

## Plate motion with explicit units

Units are never inferred: convert 5 cm/year and one million years to SI first.

```python
from earthsciencekit.plate_tectonics import plate_displacement

speed = convert(5, "cm/year", "m/s")
elapsed = convert(1e6, "year", "s")
print(f"{float(plate_displacement(speed, elapsed)) / 1e3:.0f} km")
```

## Seismic moment and magnitude

The moment magnitude and its inverse round-trip.

```python
from earthsciencekit.seismology import moment_from_magnitude, moment_magnitude

moment = moment_from_magnitude(6)
print(f"M0 = {float(moment):.3e} N m, Mw = {float(moment_magnitude(moment)):.1f}")
```

## A sampled climate history

A one-box energy balance under a 4 W/m² forcing returns an immutable `TimeSeries` with a summary, a plot and an animation.

```python
from earthsciencekit.climate import simulate_one_box

time = convert(np.linspace(0, 100, 201), "year", "s")
result = simulate_one_box(4, 1.2, 2e8, time)
print(result.summary())
figure, axes = result.plot()
```

## Invalid input is rejected

Physical bounds are checked: a temperature of 0 K raises `ValueError`.

```python
from earthsciencekit.atmosphere import dry_air_density

try:
    dry_air_density(101325, 0)
except ValueError as error:
    print(error)
```
